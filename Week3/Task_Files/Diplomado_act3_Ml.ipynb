```python
fig = px.line(data_frame=df, x="t", y="amp", width=1000, height=400,
              title='Serie de tiempo en unidades de desplazamiento',
              labels={
                     "t": "Tiempo (s)",
                     "amp": "Amplitud (m)"})
fig.show()
```

```python
#INGRESA LOS DATOS ASOCIADOS A ESTE PAR SISMOGRAMA - ESTACIÓN

A1 =  0.49 #ingresa amplitud en m
A2 = -0.32 #ingresa amplitud en m

Delta = 113.7 #ingresa distancia epicentral (km)
```

```python
A = np.abs((A1-A2)/2)*1000 #mitad de amplitud peak to peak en mm
print('variable A:', A)
```

```python
# CALCULO DE LA MAGNITUD LOCAL PARA 1 COMPONENTE, DE UNA SOLA ESTACIÓN

Ml = np.log10(A) + np.log10(Delta/100) + 3
print('Magnitud local calculada:', Ml)
```

## Solución

Para este ejercicio, se ha elegido un factor de escala para ajustar las amplitudes de los pulsos y se han definido los tiempos de liberación de momento sísmico basados en la observación de los sismogramas y la solución revisada en la actividad 1. Los valores seleccionados de amplitudes se reescalan teniendo en cuenta la solución de falla finita anterior y se calcula el mejor valor de ajuste usando un cálculo de desajuste residual (misfit):

$$\text{RMS} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} \left(d_i^{\text{obs}} - d_i^{\text{syn}}\right)^2}$$

$$\text{RMS}_{\text{promedio}} = \frac{\text{RMS}_{\text{SHEL}} + \text{RMS}_{\text{TUC}}}{2}$$

We will iterate over a range of scaling factors to find the one that minimizes the average RMS across both stations.

```python
import pandas as pd

# Tiempos de los 4 pulsos y amplitudes relativas base
d = np.array([30, 40, 45, 52])
base_amps = np.array([0.3, 0.4, 0.25, 1.0])

# Pares de señales observadas y funciones de Green / modos para cada estación
estaciones = [
    ('SHEL', shel[:500], mod[:500, 1]),
    ('TUC',  tuc[:500],  mod_tuc[:500, 1])
]

# Rango de factores a evaluar
factores_a_probar = np.arange(10, 15, 0.2)

resultados = []

for f in factores_a_probar:
    m = base_amps * f
    rms_por_estacion = {}

    for nombre, obs, modi in estaciones:
        # Matriz de diseño con las funciones retardadas
        A = np.column_stack([shift(modi, shift=di * 4, mode='nearest') for di in d])
        fwd = np.matmul(A, m)

        # Cálculo de RMS para la estación
        rms = np.sqrt(np.mean((obs - fwd)**2))
        rms_por_estacion[nombre] = rms

    rms_promedio = np.mean(list(rms_por_estacion.values()))

    resultados.append({
        'Factor': f,
        'RMS SHEL': rms_por_estacion['SHEL'],
        'RMS TUC': rms_por_estacion['TUC'],
        'RMS Promedio': rms_promedio
    })

# Creación del DataFrame
df_rms = pd.DataFrame(resultados)

# Resaltar o identificar el factor óptimo
factor_optimo = df_rms.loc[df_rms['RMS Promedio'].idxmin()]['Factor']
print(f"Factor con menor RMS promedio: {factor_optimo:.0f}\n")

# Mostrar la tabla formateada
print(df_rms.to_markdown(index=False, floatfmt=".4e"))
```

De estos resultados es posible ver que, con los pesos y tiempos de pulsos definidos, el factor que minimiza el RMS promedio es aproximadamente 11. Esto indica que las amplitudes de los pulsos deben ser escaladas por este factor para obtener la mejor coincidencia entre los sismogramas observados y sintéticos.

```python
factor = 11
amp1, amp2, amp3, amp4 = 0.3*factor,0.4*factor,0.25*factor,1*factor
t1, t2, t3, t4 =  30,40,45,52
```

3 - Si cada sismogramas sintéticos fue generado con un pulso de momento sísmico 10**19 Nm, calcule el momento sísmico y la magnitud de momento de todo el proceso de ruptura. Utilice la celda para sus cálculos

## Solución

Para calcular el momento sísmico total a partir de los pulsos de momento sísmico, se puede utilizar la siguiente relación:

$$ M_0 = \sum_{i=1}^{N} M_{0,i} = \sum_{i=1}^{N} (A_i \cdot 10^{19}\text{ Nm}) $$

donde $A_i$ son las amplitudes de los pulsos y $N$ es el número total de pulsos. En este caso, tenemos 4 pulsos con las siguientes amplitudes:

$$A_1 = 0.3 \cdot 11, \quad A_2 = 0.4 \cdot 11, \quad A_3 = 0.25 \cdot 11, \quad A_4 = 1.0 \cdot 11$$

Por lo tanto, el momento sísmico total se calcula como:

```python
import math

def calcular_momento_sismico(amplitudes, momento_por_pulso=1e19):
    """
    Calcula el momento sísmico total a partir de las amplitudes de los pulsos y el momento por pulso.
    """
    momento_total = sum(amplitudes) * momento_por_pulso
    return momento_total

def calcular_mw(m0):
    """
    Calcula la magnitud de momento (Mw) a partir del momento sísmico (M0).
    M0 debe estar expresado en Newton-metro (N·m).
    """
    if m0 <= 0:
        raise ValueError("El momento sísmico (M0) debe ser un número positivo.")

    mw = (2 / 3) * math.log10(m0) - 6.06
    return round(mw, 2)

momento_sismico = calcular_momento_sismico(m)
magnitud = calcular_mw(momento_sismico)

print(f"La magnitud de momento (Mw) a partir de un momento sísmico de {momento_sismico} N·m es: {magnitud}")
```

Obtenemos un valor de magnitud de momento (Mw) de aproximadamente 7.49, lo cual es consistente con la magnitud reportada por el USGS para este evento sísmico (7.41). Esto confirma que la modelación de la fuente sísmica mediante los pulsos seleccionados y sus respectivas amplitudes es adecuada para representar el proceso de ruptura del terremoto de Colombia del 10 de Agosto de 2026.

### Gráfico función temporal

Para este apartado, haré uso de un formato de gráfico estilo USGS, que es característico por su claridad y presentación profesional. A continuación, se presenta la función temporal de la fuente sísmica modelada como una secuencia de pulsos, junto con un relleno bajo la curva y una línea discontinua que indica el término del proceso de ruptura.

```python
# -------------------------------------------------------------
# 1. Definición y cálculo de la función de tasa de momento
# -------------------------------------------------------------
d = np.array([30, 40, 45, 52])
factor = 11
m = [0.3 * factor, 0.4 * factor, 0.25 * factor, 1.0 * factor]

t = np.arange(0, 500) * 0.25
y = np.zeros_like(t)

for mi, di in zip(m, d):
    centro, ancho = di, 7
    y += np.maximum(1.0 - np.abs(t - centro) / (ancho / 2.0), 0.0) * mi

# Normalización de la curva a tasa relativa (0 a 1)
y_rel = y / np.max(y)

# -------------------------------------------------------------
# 2. Configuración estética estilo USGS
# -------------------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 6.5), dpi=150)

# Curva principal y relleno gris bajo la curva
ax.plot(t, y_rel, color='black', linewidth=2.2, zorder=3)
ax.fill_between(t, y_rel, color='#e5e5e5', zorder=2)

# Línea discontinua roja de término del proceso de ruptura (~55 s)
ax.axvline(x=55.5, color='red', linestyle='--', linewidth=2.0, zorder=4)

# Formato y estilo de los ejes
ax.set_xlim(0, 90)
ax.set_ylim(0.0, 1.0)
ax.set_xlabel('Time (s)', fontsize=14, labelpad=8)
ax.set_ylabel('Relative Moment Rate (Nm/s)', fontsize=14, labelpad=8)

# Borde exterior grueso y marcas de graduación (ticks) hacia adentro
for spine in ax.spines.values():
    spine.set_linewidth(2.5)
    spine.set_color('black')

ax.minorticks_on()
ax.tick_params(which='major', direction='in', length=6, width=1.8, top=True, right=True, labelsize=12)
ax.tick_params(which='minor', direction='in', length=3, width=1.0, top=True, right=True)

# Cuadrícula punteada tenue
ax.grid(True, which='major', linestyle=':', color='gray', alpha=0.7, zorder=1)

# Cuadro de texto informativo en la esquina superior derecha
info_text = (
    r"$\mathbf{M_0:\ 2.145E+20\ Nm}$" + "\n" +
    r"$\mathbf{M_w:\ 7.49}$"
)
ax.text(
    0.98, 0.97, info_text,
    transform=ax.transAxes,
    fontsize=13,
    verticalalignment='top',
    horizontalalignment='right',
    zorder=5
)

plt.tight_layout()
plt.show()
```
